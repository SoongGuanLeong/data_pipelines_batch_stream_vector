## Load Configs

```python
cfg = load_config()

JOB_NAMES = cfg["spark_jobs"]["jobs"]
CATALOG = cfg["general"]["catalog"]
SILVER_NAMESPACE = cfg["general"]["namespaces"]["silver"]
GOLD_NAMESPACE = cfg["general"]["namespaces"]["gold"]

cfg_cust= cfg["gold"]["dim_customers_scd2"]
TARGET_TABLE = cfg_cust["target_table"]
SOURCE_TABLE = cfg_cust["source_table"]
GEO_TABLE = cfg_cust["geo_table"]
BUFFER_HOURS = cfg_cust["buffer_hours"]
KEY_COLUMN = cfg_cust["key_column"]
```

## Import Libraries and Start Session

```python
from pyspark.sql import SparkSession
from pyspark.sql import functions as F
from pyspark.sql import Window as W
import pyspark
import datetime
import json

spark = (
    SparkSession.builder
        .appName(JOB_NAMES["gold"])
        .getOrCreate()
)
```

## Run Pipeline

```python
def run_dim_customers_pipeline(spark):

    last_commit_ts = get_last_commit_ts(spark, TARGET_TABLE)

    if last_commit_ts is None:
        dim_customers_df = build_dim_customers_scd2(
            spark,
            SOURCE_TABLE,
            GEO_TABLE,
        )

        overwrite_table(dim_customers_df, TARGET_TABLE)
        return

    effective_ts = get_effective_watermark(
        last_commit_ts,
        BUFFER_HOURS,
    )

    changed_customer_ids = get_changed_keys(
        spark,
        SOURCE_TABLE,
        KEY_COLUMN,
        effective_ts,
    )

    if changed_customer_ids.limit(1).count() == 0:
        return

    incremental_dim_df = build_incremental_dim_customers(
        spark,
        SOURCE_TABLE,
        GEO_TABLE,
        changed_customer_ids,
    )

    replace_by_key(
        spark,
        incremental_dim_df,
        TARGET_TABLE,
        KEY_COLUMN,
    )
        
```
