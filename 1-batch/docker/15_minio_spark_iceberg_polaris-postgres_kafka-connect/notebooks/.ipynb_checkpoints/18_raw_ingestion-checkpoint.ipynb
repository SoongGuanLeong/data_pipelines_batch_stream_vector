## Import Libraries and Start Session

```python
from pyspark.sql import SparkSession
from pyspark.sql import functions as F
from pyspark.sql import Window as W
from pyspark.sql.types import *
import pyspark
import datetime
import json

spark = (
    SparkSession.builder
        .appName(JOB_NAMES["raw_ingestion"])
        .getOrCreate()
)
```

## DDL: raw_ingestion_metadata

```python
table_name = f"{CATALOG}.{RAW_FOLDER}.{METADATA_TABLE}"

spark.sql(f"""
    CREATE TABLE IF NOT EXISTS {table_name} (
        file_name STRING,
        file_type STRING,
        path STRING,
        size_bytes BIGINT,
        upload_ts TIMESTAMP,
        checksum STRING
    )
    USING ICEBERG
    PARTITIONED BY (days(upload_ts))
""")

print(f"{table_name} created.")
```

## Function: Upload Raw File

```python
def upload_raw_file(file_path:str):
    """Upload file to MinIO and log metadata to Iceberg"""
    
    file_path = Path(file_path)
    if not file_path.exists():
        raise FileNotFoundError(f"{file_path} does not exist")

    # dedup check
    file_checksum = checksum(file_path)

    exists = (
        spark.table(table_name)
            .filter(F.col("checksum") == file_checksum)
            .limit(1)
            .count()
    )

    if exists:
        print("File already ingested. Skipping metadata write.")
        return spark_path

    # Upload via mc
    target_path = get_target_path(file_path)

    subprocess.run(
        ["mc", "cp", str(file_path), target_path],
        check=True
    )

    # metadata
    file_size = file_path.stat().st_size
    upload_ts = datetime.utcnow()
    file_type = safe_file_type(file_path)
    spark_path = get_spark_path(file_path)

    metadata_df = spark.createDataFrame([{
        "file_name": file_path.name,
        "file_type": file_type,
        "path": spark_path,
        "size_bytes": file_size,
        "upload_ts": upload_ts,
        "checksum": file_checksum
    }])

    metadata_df.write.format("iceberg").mode("append").saveAsTable(table_name)

    print(f"[{upload_ts}] Uploaded {file_path.name} → {spark_path} ({file_size} bytes, checksum={file_checksum})")
    return spark_path
```

```python
from pathlib import Path

# ----------------------------
# Path inside the container
# ----------------------------
RAW_DATA_DIR = Path("/opt/spark/datasets")

files = list(RAW_DATA_DIR.glob("*"))

if not files:
    print("No files found.")
else:
    uploaded_paths = []

    for file in files:
        spark_path = upload_raw_file(file)
        uploaded_paths.append(spark_path)

    print("\nUploaded paths:")
    for p in uploaded_paths:
        print(p)
```
