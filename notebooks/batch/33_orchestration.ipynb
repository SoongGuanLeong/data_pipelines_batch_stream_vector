```python
from nbformat import read
from nbclient import NotebookClient
import sys


def run_notebook(path):
    try:
        with open(path) as f:
            nb = read(f, as_version=4)

        client = NotebookClient(nb, timeout=None, kernel_name="python3")
        client.execute()

        print(f"SUCCESS: {path}")

    except Exception as e:
        print(f"FAILED: {path}")
        raise e
```

```python
notebooks = [
    "18_init_lakehouse.ipynb",
    "19_raw_ingestion.ipynb",
    "20_bronze_cdc_ddl.ipynb",
    "21_bronze_cdc_ingestion.ipynb",
    "22_bronze_lookup_ddl_ingestion.ipynb",
    "23_silver.ipynb",
    "24_gold_dim_customers.ipynb",
    "25_gold_dim_sellers.ipynb",
    "26_gold_dim_products.ipynb",
    "27_gold_dim_date.ipynb",
    "28_gold_dims_bootstrap.ipynb",
    "29_gold_fact_orders.ipynb",
    "30_gold_fact_order_items.ipynb",
    "31_gold_fact_order_payments.ipynb",
    "32_gold_fact_order_reviews.ipynb",
]
```

```python
for nb in notebooks:
    run_notebook(nb)
```
